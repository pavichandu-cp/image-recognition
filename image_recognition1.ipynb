```python
from transformers import pipeline

print("Loading Bird Recognition AI...")

classifier = pipeline(
    "image-classification",
    model="chriamue/bird-species-classifier"
)

print("Bird recognition model loaded successfully!")
```

```python
import gradio as gr
from PIL import Image

def recognize_bird(image):

    if image is None:
        return "Please upload a bird image first!"

    image = image.convert("RGB")

    # Predict bird species
    predictions = classifier(image, top_k=5)

    best = predictions[0]

    bird_name = best["label"].replace("_", " ").title()
    confidence = best["score"] * 100

    response = f"""
🐦 BIRD RECOGNITION RESULT

Identified Bird: {bird_name}

Confidence: {confidence:.2f}%

TOP 5 BIRD PREDICTIONS:
"""

    for item in predictions:
        name = item["label"].replace("_", " ").title()
        score = item["score"] * 100

        response += f"\n• {name}: {score:.2f}%"

    response += """

Note: AI predictions may not always be correct.
"""

    return response
```

```python
with gr.Blocks(theme=gr.themes.Soft()) as app:

    gr.Markdown("""
    # 🐦 Bird Recognition AI
    ### Intelligent Bird Image Classification

    Upload a bird photograph and discover its species!
    """)

    with gr.Row():

        with gr.Column():

            image_input = gr.Image(
                type="pil",
                label="Upload Bird Photograph"
            )

            recognize_button = gr.Button(
                "🔍 Recognize Bird",
                variant="primary"
            )

        with gr.Column():

            result_output = gr.Textbox(
                label="Bird Recognition Results",
                lines=14,
                interactive=False
            )

    recognize_button.click(
        fn=recognize_bird,
        inputs=image_input,
        outputs=result_output
    )

    gr.Markdown("""
    ### How to use

    1. Upload a bird image.
    2. Click Recognize Bird.
    3. View the identified species.
    4. Check the top five predictions.

    **Developed using Python, Hugging Face, PyTorch and Gradio.**
    """)

app.launch(share=True)
```
